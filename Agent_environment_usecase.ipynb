```python
# ==========================================
# Online Shopping Multi-Agent System
# With Orchestration
# ==========================================


# ------------------------------------------
# Agent 1 : Product Search Agent
# ------------------------------------------

def search_agent(product):

    product = product.strip().lower()

    products = {
        "laptop": ["HP Laptop", "Dell Laptop"],
        "phone": ["Samsung Phone", "iPhone"],
        "headphone": ["Boat Headphone", "Sony Headphone"]
    }

    if product in products:

        print("\n--- Search Agent ---")
        print("Products Found:")

        for item in products[product]:
            print("-", item)

        # Selecting first product
        return products[product][0]

    else:
        print("\nProduct not found")
        return None



# ------------------------------------------
# Agent 2 : Price Comparison Agent
# ------------------------------------------

def price_agent(product):

    prices = {

        "HP Laptop": 50000,
        "Dell Laptop": 55000,

        "Samsung Phone": 25000,
        "iPhone": 70000,

        "Boat Headphone": 2000,
        "Sony Headphone": 4000
    }


    print("\n--- Price Comparison Agent ---")

    if product in prices:

        price = prices[product]

        print("Selected Product :", product)
        print("Best Price :", price)

        return price

    else:
        print("Price not available")
        return None



# ------------------------------------------
# Agent 3 : Payment Agent
# ------------------------------------------

def payment_agent(product, price):

    print("\n--- Payment Agent ---")

    print("Processing Payment...")

    print("Product :", product)

    print("Amount :", price)

    print("Payment Successful")

    print("Order Confirmed")



# ------------------------------------------
# Orchestrator Agent
# ------------------------------------------

def orchestrator():

    print("==============================")
    print("Online Shopping Multi-Agent System")
    print("==============================")


    user_input = input(
        "Enter Product Category (Laptop/Phone/Headphone): "
    )


    # Calling Search Agent

    selected_product = search_agent(user_input)


    if selected_product is not None:


        # Calling Price Agent

        price = price_agent(selected_product)


        if price is not None:


            # Calling Payment Agent

            payment_agent(
                selected_product,
                price
            )


            print("\nWorkflow Completed Successfully")


    else:

        print("\nWorkflow Stopped")



# ------------------------------------------
# Start Application
# ------------------------------------------

orchestrator()
```
